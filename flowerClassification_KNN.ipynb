```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report

```

```python
iris=load_iris()
```

```python
x = iris.data
feature_names = iris.feature_names
y = iris.target
target_names = iris.target_names
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    x, y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

```python
print(x)
print(y)
```

```python
model=make_pipeline(
    StandardScaler(),
    KNeighborsClassifier(n_neighbors=5, metric="euclidean")
)
```

```python
model.fit(X_train, y_train)
```

```python
y_pred=model.predict(X_test)
```

```python
print(f"y_pred for {X_test} are {y_pred}")
```

```python
print(f"KNN Model Accuracy(k=5): {accuracy_score(y_pred, y_test)*100}")
```

```python
new_flower=[[
    5, 18.5, 37, 1.3
]]
```

```python
predicted_class_idx=model.predict(new_flower)[0]
print(predicted_class_idx)
```

```python
print(f"Predicted Species:{target_names[predicted_class_idx]}")
```
